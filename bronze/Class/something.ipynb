```python
N = int(input())#A 길이랄까 6
A = list(map(int, input().split()))#숫자열 1 2 3 4 4 4
 
cnt_left = [0] * (N + 1)# [0,0,0,0,0,0,0]
cnt_right = [0] * (N + 1)#[0,0,0,0,0,0,0]
for x in A:# 1 2 3 4 4 4를 한번씩 돌아요
    cnt_right[x] += 1# x=3이면 세번째 0에 1을 더함 (cnt_right의 [4]가 3이라면 A에 4가 세개 있다는 뜻)
    #cnt_right=[0,1,1,1,3]
 
ans = 0
distinct_left = 0#내가 지금 보고있는 숫자 왼쪽에있는 숫자의 종류수
 
for i in range(N):# N=6
    if cnt_right[A[i]] == 2:#A[i] 오른쪽에 특정 알파벳이 2개라면
        ans += distinct_left - (cnt_left[A[i]] > 0)#- (cnt_left[A[i]] > 0)는 같은 종류의 숫자가 겹친다면 그대로 뺀다. MOO는 되지만 OOO는 안되기 때문. 왼쪽에 있는 숫자의 종류만큼 답에 더한다
        
    cnt_right[A[i]] -= 1 #방금 cnt_right[A[i]]를 확인했으니 지워야 한다. 다음에 나올 cnt_right[A[i+1]]에게는 cnt_right[A[i]]가 더이상 오른쪽에 있는 숫자가 아니기 때문
    distinct_left += cnt_left[A[i]] == 0 #왼쪽에서 한번도 못본 숫자이면 distinct_left를 1 증가
    cnt_left[A[i]] += 1#다음칸으로 이동하면 cnt_right[A[i]]는 왼쪽에 있는 숫자의 종류에 포함이 됨.
 
print(ans)
```

```python
#MooHunt Solution
import sys
input = sys.stdin.readline

N, K = map(int, input().split())#N:보드길이
moves = []
for _ in range(K):
    x, y, z = map(int, input().split())
    moves.append((x - 1, y - 1, z - 1))  # 인덱스는 0부터 시작하니깐

max_score = 0
count = 0

board = [''] * N#빈보드

def dfs(pos):
    global max_score, count

    if pos == N:#pos=0 에서 시작
        score = 0
        for x, y, z in moves:#x-1 ,y-1 ,z-1을 돌아요
            if board[x] == 'M' and board[y] == 'O' and board[z] == 'O':#점수판이 가리키는게 M,O,O 이면 점수 1점추가
                score += 1

        if score > max_score:#최고점수 갱신
            max_score = score#최고점수 하나일때
            count = 1#최고점수는 한개
        elif score == max_score:#공동1등일때
            count += 1#최고점수 개수 += 1
        return

    board[pos] = 'M'#board[0]='M'
    dfs(pos + 1)    #pos(1)

    board[pos] = 'O'
    dfs(pos + 1)

dfs(0)
print(max_score, count)
```

```python
#purchasing milk
import sys
input = sys.stdin.readline

N, Q = map(int, input().split())# N = deal의 개수 / Q = 사고싶은 우유 개수
a = list(map(int, input().split()))# 가격(deal) 의 종류

for i in range(1, N):
    a[i] = min(a[i], 2 * a[i - 1]) # a[i]보다 뒤에 있는 거(a[i-1])는 우유를 a[i]보다 두배나 적게 판다. 파는 양은 a[i] == 2* a[i-1]임. 그래서 가격이 더 싼 친구는 냄겨놓고 비싼애는 버려야됨.

sizes = [1]#싸이즈?!
for i in range(1, N):
    sizes.append(sizes[-1] * 2)#deal당 파는 양 [2,4,8,16,32]이렇게 2의 배수로 늘어남

for _ in range(Q):#질문 개수
    x = int(input())
    need = x#    x = 사고싶은 우유의 양
    cost = 0
    answer = float('inf') 

    for i in range(N - 1, -1, -1):
        if sizes[i] <= need:
            cnt = need // sizes[i]
            cost += cnt * a[i]
            need -= cnt * sizes[i]

        if need > 0:
            answer = min(answer, cost + a[i])
        else:
            answer = min(answer, cost)

    print(answer)
```

```python
#Load Balancing
with open("balancing.in", "r") as fin:
    line1 = fin.readline().split()
    if not line1: exit()
    cowsnum, farm = map(int, line1)
    
    coords = []
    x_coords = []
    y_coords = []
    
    for i in range(cowsnum):
        a = list(map(int, fin.readline().split()))
        coords.append(a)
        # 소의 좌표 옆(짝수)에 울타리를 쳐야 하므로 후보군 저장
        x_coords.append(a[0] + 1)
        y_coords.append(a[1] + 1)

anslist=[]
for xfence in range(farm):
    for yfence in range(farm):
        F=[0,0,0,0,0]
        for coord in coords: #[x,y]
            if coord[0]>xfence:#오른쪽
                if coord[1]>yfence:#위에
                    F[1]+=1 #제1사분면에 1 더하기
                else:#오른쪽 아래에 (4사분면)
                    F[4]+=1
            else:#왼쪽
                if coord[1]>yfence:#위에
                    F[2]+=1 #제2사분면에 1 더하기
                else:
                    F[3]+=1 #제 3사분면
        anslist.append(max(F))

with open("balancing.out", "w") as fout:
    fout.write(str(min(anslist)) + "\n")
```
